```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from keras.models import Sequential
from keras.layers import LSTM, Dense, Dropout
from sklearn.metrics import mean_squared_error
from keras.callbacks import EarlyStopping
```

```python
data = pd.read_csv('DFF.csv')
```

```python
# Affichage des premières lignes pour vérification
print(data.head())
```

```python
# Conversion de la colonne Date en format datetime
data['Date'] = pd.to_datetime(data['Date'])
```

```python
# Trier par date si ce n'est pas déjà fait
data = data.sort_values('Date')
```

```python
# Vérification des valeurs manquantes
print(data.isnull().sum())
```

```python
# Remplir les valeurs manquantes par la moyenne de chaque colonne (solution simple)
data = data.fillna(data.mean())
```

```python
# Renommer les colonnes pour simplifier (facultatif)
data = data.rename(columns={
    'Sol Rad (W/sq.m)': 'sol_rad',
    'Net Rad (W/sq.m)': 'net_rad',
    'Avg Vap Pres (kPa)': 'vap_pres',
    'Avg Air Temp (C)': 'air_temp',
    'Avg Rel Hum (%)': 'rel_hum',
    'Avg Wind Speed (m/s)': 'wind_speed',
    'Wind Run (km)': 'wind_run',
    'Avg Soil Temp (C)': 'soil_temp',
    'PM ETo (mm)': 'eto'
})
```

```python
# Colonnes à normaliser
features = ['sol_rad', 'net_rad', 'vap_pres', 'air_temp', 'rel_hum', 'wind_speed', 'wind_run', 'soil_temp']
scaler = StandardScaler()
data[features] = scaler.fit_transform(data[features])
```

```python
# Ajouter la colonne cible à la fin (juste pour bien structurer)
cols = features + ['eto']
data = data[cols]
```

```python
# 4. Créer des séquences temporelles
def create_sequences(data, sequence_length=5):
    sequences = []
    labels = []
    for i in range(len(data) - sequence_length):
        seq = data.iloc[i:i+sequence_length].drop('eto', axis=1).values
        label = data.iloc[i+sequence_length]['eto']
        sequences.append(seq)
        labels.append(label) 
    return np.array(sequences), np.array(labels)
```

```python
X, y = create_sequences(data)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
```

```python
model = Sequential()
model.add(LSTM(units=64, return_sequences=False, input_shape=(X_train.shape[1], X_train.shape[2])))
model.add(Dropout(0.2))
model.add(Dense(1))
# EarlyStopping pour éviter le sur-apprentissage
early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
model.compile(optimizer='adam', loss='mean_squared_error')
```

```python
y_pred = model.predict(X_test)
```

```python
model.fit(X_train, y_train, epochs=50, batch_size=32, validation_data=(X_test, y_test), callbacks=[early_stop])
```

```python
# Affichage des courbes de perte pour analyse
plt.figure(figsize=(8, 4))
plt.plot(model.history.history['loss'], label='Entraînement')
plt.plot(model.history.history['val_loss'], label='Validation')
plt.title("Courbe de perte (Loss)")
plt.xlabel("Époques")
plt.ylabel("Loss (MSE)")
plt.legend()
plt.show()
```

```python
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse}")
```

```python
plt.figure(figsize=(10,6))
plt.plot(y_test, label='Valeur réelle')
plt.plot(y_pred, label='Prédiction')
plt.legend()
plt.title("Prédiction des besoins en irrigation (PM ETo)")
plt.xlabel("Temps")
plt.ylabel("ETo (normalisé)")
plt.tight_layout()
plt.show()
```
